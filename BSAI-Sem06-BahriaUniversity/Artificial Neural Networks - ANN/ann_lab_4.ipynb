**Aim:** Implement and compare different activation functions (Sigmoid, Tanh, ReLU, Leaky ReLU) used in neural networks.

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

x = np.linspace(-10, 10, 100)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(x, sigmoid(x))
plt.title('Sigmoid Activation Function')
plt.grid(True)

plt.subplot(2, 2, 2)
plt.plot(x, tanh(x))
plt.title('Tanh Activation Function')
plt.grid(True)

plt.subplot(2, 2, 3)
plt.plot(x, relu(x))
plt.title('ReLU Activation Function')
plt.grid(True)

plt.subplot(2, 2, 4)
plt.plot(x, leaky_relu(x))
plt.title('Leaky ReLU Activation Function')
plt.grid(True)

plt.tight_layout()
plt.show()
```

## 2. Derivatives of activation functions

The derivative of each function determines how gradients flow during backpropagation.

```python
def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

def tanh_derivative(x):
    return 1 - np.tanh(x) ** 2

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

def leaky_relu_derivative(x, alpha=0.01):
    return np.where(x > 0, 1, alpha)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(x, sigmoid_derivative(x))
plt.title('Sigmoid Derivative')
plt.grid(True)

plt.subplot(2, 2, 2)
plt.plot(x, tanh_derivative(x))
plt.title('Tanh Derivative')
plt.grid(True)

plt.subplot(2, 2, 3)
plt.plot(x, relu_derivative(x))
plt.title('ReLU Derivative')
plt.grid(True)

plt.subplot(2, 2, 4)
plt.plot(x, leaky_relu_derivative(x))
plt.title('Leaky ReLU Derivative')
plt.grid(True)

plt.tight_layout()
plt.show()
```

## 3. Analysis of results

**Sigmoid:** Squashes inputs into the range (0, 1). Saturates for large positive or negative inputs, which drives its derivative toward zero and slows learning in deep networks (vanishing gradient).

**Tanh:** Zero-centered output in the range (-1, 1), which generally gives better gradient flow than sigmoid, but still saturates at the extremes.

**ReLU:** Computationally cheap and avoids saturation for positive inputs, but neurons can become inactive ("dying ReLU") when inputs stay negative, since the gradient is zero there.

**Leaky ReLU:** Addresses the dying ReLU problem by allowing a small, non-zero gradient (controlled by alpha) when the input is negative.

## 4. Conclusion

The choice of activation function affects both the speed of convergence and the stability of training. Sigmoid and tanh are more prone to vanishing gradients in deep networks, while ReLU and its variants like Leaky ReLU are generally preferred for hidden layers because they mitigate this issue and are cheaper to compute. The correct choice depends on the network architecture and the specific problem being solved.
